# Performing NumPy Pandas on a DataSet

```python
import pandas as pd
import numpy as np

df = pd.read_csv("/Chocolate Sales (2).csv")
```

# DATA CLEANING


Convert Amount to numeric


```python
df['Amount'] = (
    df['Amount']
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
)
```

Convert Date to datetime

```python
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)
```

Check missing values

```python
df.isnull().sum()
```

# Pandas Operations

Select Columns

```python
df['Country']
df[['Country', 'Amount']]

```

Filter Rows

```python
# Sales in India
df[df['Country'] == 'India']

# Sales above 10,000
df[df['Amount'] > 10000]
```

Sort Data

```python
df.sort_values(by='Amount', ascending=False)
```

# GroupBy (Most Important in Pandas)

Total Sales by Country

```python
df.groupby('Country')['Amount'].sum()
```

Average Sales by Product

```python
df.groupby('Product')['Amount'].mean()
```

Total Boxes Shipped per Sales Person

```python
df.groupby('Sales Person')['Boxes Shipped'].sum()
```

# Date & Time Analysis

Extract Year, Month

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
#print(df)
```

Monthly Sales

```python
df.groupby('Month')['Amount'].sum()
```

# Value Counts (Quick Analysis)

```python
df['Country'].value_counts()
df['Product'].value_counts()
```

# Simple Business Insights

Top 5 Products by Revenue

```python
df.groupby('Product')['Amount'].sum().sort_values(ascending=False).head(5)
```

Best Sales Person

```python
df.groupby('Sales Person')['Amount'].sum().idxmax()
```
